### 🧹 1. Limpeza dos Dados
**Introdução:** Antes de analisar qualquer dado, precisamos garantir que ele esteja íntegro. Dados ausentes ou corrompidos podem distorcer nossos resultados.
**Descrição do Código:** No bloco abaixo, vamos importar as bibliotecas essenciais (`pandas`, `matplotlib` e `seaborn`), carregar o dataset dos pinguins e remover qualquer linha que contenha valores nulos (NaN), garantindo um conjunto de dados limpo para trabalharmos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração de estilo para os gráficos
sns.set_theme(style="whitegrid")

# Carregando o dataset diretamente via seaborn (disponível no Kaggle)
df = sns.load_dataset('penguins')

print("📦 Formato original do dataset:", df.shape)

# Limpeza: Removendo linhas com valores nulos (NaN)
df_clean = df.dropna()

print("✨ Formato após limpeza dos dados:", df_clean.shape)
display(df_clean.head())
```

### 📊 2. Análise Exploratória
**Introdução:** Com os dados limpos, o próximo passo é entender o perfil matemático do nosso dataset e como as categorias estão divididas.
**Descrição do Código:** A seguir, utilizaremos a função `.describe()` para extrair um resumo estatístico numérico (média, desvio padrão, valores máximos e mínimos). Em seguida, faremos uma contagem com `.value_counts()` para descobrir exatamente quantos pinguins de cada espécie temos na nossa base.

```python
# Resumo estatístico numérico
display(df_clean.describe())

# Verificando a distribuição das espécies
print("\n🐧 Contagem de observações por Espécie:")
print(df_clean['species'].value_counts())
```

### 📈 3. Visualizações Informativas
**Introdução:** Números sozinhos podem ser difíceis de interpretar. A visualização de dados nos ajuda a encontrar padrões ocultos de forma intuitiva.
**Descrição do Código:** Criaremos dois gráficos lado a lado:
1. **Boxplot (Gráfico de Caixa):** Mostrará como a massa corporal varia dentro de cada espécie, destacando a mediana e possíveis *outliers* (valores atípicos).
2. **Gráfico de Dispersão (Scatterplot):** Cruzará o comprimento do bico com a massa corporal, colorindo os pontos por espécie para verificarmos se existe uma correlação visual entre essas medidas.

```python
plt.figure(figsize=(15, 6))

# Gráfico 1: Boxplot (Distribuição de peso por espécie)
plt.subplot(1, 2, 1)
sns.boxplot(data=df_clean, x='species', y='body_mass_g', palette='Set2')
plt.title('📦 Distribuição de Massa Corporal por Espécie', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Espécie', fontsize=12)
plt.ylabel('Massa Corporal (g)', fontsize=12)

# Gráfico 2: Scatterplot (Relação entre Bico e Peso)
plt.subplot(1, 2, 2)
sns.scatterplot(data=df_clean, x='bill_length_mm', y='body_mass_g', hue='species', palette='Set2', alpha=0.8, s=80)
plt.title('🔸 Comprimento do Bico vs. Massa Corporal', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Comprimento do Bico (mm)', fontsize=12)
plt.ylabel('Massa Corporal (g)', fontsize=12)

plt.tight_layout()
plt.show()
```

### 🔍 4. Identificação de Insights e 📋 5. Conclusões

**Principais Insights Encontrados:**
* 🚀 **Diferença Fisiológica:** Analisando o Boxplot, fica evidente que a espécie *Gentoo* é significativamente mais pesada que as espécies *Adelie* e *Chinstrap*, apresentando uma faixa de massa corporal bem distinta.
* 📈 **Correlação Positiva:** O gráfico de dispersão revela uma tendência clara: pinguins com bicos mais longos tendem a ter uma massa corporal maior, indicando um padrão de crescimento proporcional.
* 🎨 **Clusters (Agrupamentos) Naturais:** As características mapeadas formam agrupamentos bem definidos por cor no gráfico de dispersão, comprovando que as medidas físicas são excelentes diferenciadores de espécie.
